## CELL 2 — Imports

```python
import os

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

## CELL 3 — Load Results CSV Files

**Expected Output:** Two tables containing Seed, Mean Episode Reward, Mean Burned Cells, Mean Fire Intensity.

```python
RESULT_DIR = "/content/drive/MyDrive/PyroRL_Saudi_Project/results"

saudi_df = pd.read_csv(
    f"{RESULT_DIR}/multi_seed_results_saudi.csv"
)

california_df = pd.read_csv(
    f"{RESULT_DIR}/multi_seed_results_california.csv"
)

print("Saudi Results:")
display(saudi_df)

print("\nCalifornia Results:")
display(california_df)
```

## CELL 4 — Compute Final Statistics

```python
summary_df = pd.DataFrame({

    "Environment": [

        "Saudi",
        "California"
    ],

    "Mean Reward": [

        saudi_df["Mean Episode Reward"].mean(),

        california_df["Mean Episode Reward"].mean()
    ],

    "Reward Std": [

        saudi_df["Mean Episode Reward"].std(),

        california_df["Mean Episode Reward"].std()
    ],

    "Mean Burned Cells": [

        saudi_df["Mean Burned Cells"].mean(),

        california_df["Mean Burned Cells"].mean()
    ],

    "Burned Cells Std": [

        saudi_df["Mean Burned Cells"].std(),

        california_df["Mean Burned Cells"].std()
    ],

    "Mean Fire Intensity": [

        saudi_df["Mean Fire Intensity"].mean(),

        california_df["Mean Fire Intensity"].mean()
    ],

    "Fire Intensity Std": [

        saudi_df["Mean Fire Intensity"].std(),

        california_df["Mean Fire Intensity"].std()
    ]
})

summary_df
```

## CELL 5 — Save Summary Table

```python
summary_path = f"{RESULT_DIR}/final_summary_statistics.csv"

summary_df.to_csv(
    summary_path,
    index=False
)

print(f"Saved summary table:\n{summary_path}")
```

## CELL 6 — Main Reward Comparison Figure
*Note: This is the MAIN PAPER FIGURE. Expected: Saudi has better containment, California has harder wildfire dynamics.*

```python
plt.figure(figsize=(8,5))

envs = summary_df["Environment"]

mean_rewards = summary_df["Mean Reward"]

reward_stds = summary_df["Reward Std"]

plt.bar(
    envs,
    mean_rewards,
    yerr=reward_stds,
    capsize=8
)

plt.ylabel("Mean Episode Reward")

plt.title(
    "Cross-Regional PPO Reward Comparison"
)

plt.grid(axis='y')

plt.show()
```

## CELL 7 — Burned Cells Error-Bar Plot

```python
plt.figure(figsize=(8,5))

mean_burn = summary_df["Mean Burned Cells"]

burn_std = summary_df["Burned Cells Std"]

plt.bar(
    envs,
    mean_burn,
    yerr=burn_std,
    capsize=8
)

plt.ylabel("Mean Burned Cells")

plt.title(
    "Wildfire Burned Area Comparison"
)

plt.grid(axis='y')

plt.show()
```

## CELL 8 — Fire Intensity Error-Bar Plot

```python
plt.figure(figsize=(8,5))

mean_fire = summary_df["Mean Fire Intensity"]

fire_std = summary_df["Fire Intensity Std"]

plt.bar(
    envs,
    mean_fire,
    yerr=fire_std,
    capsize=8
)

plt.ylabel("Mean Fire Intensity")

plt.title(
    "Final Fire Intensity Comparison"
)

plt.grid(axis='y')

plt.show()
```

## CELL 9 — Reward Distribution Boxplot

```python
plt.figure(figsize=(8,5))

plt.boxplot([

    saudi_df["Mean Episode Reward"],

    california_df["Mean Episode Reward"]

],

labels=[

    "Saudi",
    "California"
])

plt.ylabel("Mean Episode Reward")

plt.title(
    "Reward Distribution Across Seeds"
)

plt.grid(axis='y')

plt.show()
```

## CELL 10 — Burned Cells Boxplot

```python
plt.figure(figsize=(8,5))

plt.boxplot([

    saudi_df["Mean Burned Cells"],

    california_df["Mean Burned Cells"]

],

labels=[

    "Saudi",
    "California"
])

plt.ylabel("Mean Burned Cells")

plt.title(
    "Burned Cells Distribution Across Seeds"
)

plt.grid(axis='y')

plt.show()
```

## CELL 11 — Fire Intensity Boxplot

```python
plt.figure(figsize=(8,5))

plt.boxplot([

    saudi_df["Mean Fire Intensity"],

    california_df["Mean Fire Intensity"]

],

labels=[

    "Saudi",
    "California"
])

plt.ylabel("Mean Fire Intensity")

plt.title(
    "Fire Intensity Distribution Across Seeds"
)

plt.grid(axis='y')

plt.show()
```

## CELL 12 — Transfer Degradation Calculation

```python
saudi_reward = (
    saudi_df["Mean Episode Reward"].mean()
)

california_reward = (
    california_df["Mean Episode Reward"].mean()
)

transfer_gap = (
    california_reward - saudi_reward
)

relative_degradation = (

    abs(transfer_gap)
    /
    abs(saudi_reward)

) * 100

print(f"Transfer Gap: {transfer_gap:.2f}")

print(
    f"Relative Degradation: "
    f"{relative_degradation:.2f}%"
)
```

## CELL 13 — Transfer Degradation Figure

```python
plt.figure(figsize=(8,5))

labels = [

    "Saudi Native",
    "California"
]

values = [

    saudi_reward,
    california_reward
]

plt.bar(labels, values)

plt.ylabel("Mean Episode Reward")

plt.title(
    "Cross-Regional Transfer Degradation"
)

plt.grid(axis='y')

plt.show()
```

## CELL 14 — Publication Table

```python
publication_table = pd.DataFrame({

    "Evaluation": [

        "Saudi Native PPO",
        "California PPO"
    ],

    "Reward Mean ± Std": [

        f"{summary_df.iloc[0]['Mean Reward']:.2f} ± "
        f"{summary_df.iloc[0]['Reward Std']:.2f}",

        f"{summary_df.iloc[1]['Mean Reward']:.2f} ± "
        f"{summary_df.iloc[1]['Reward Std']:.2f}"
    ],

    "Burned Cells Mean ± Std": [

        f"{summary_df.iloc[0]['Mean Burned Cells']:.2f} ± "
        f"{summary_df.iloc[0]['Burned Cells Std']:.2f}",

        f"{summary_df.iloc[1]['Mean Burned Cells']:.2f} ± "
        f"{summary_df.iloc[1]['Burned Cells Std']:.2f}"
    ],

    "Fire Intensity Mean ± Std": [

        f"{summary_df.iloc[0]['Mean Fire Intensity']:.2f} ± "
        f"{summary_df.iloc[0]['Fire Intensity Std']:.2f}",

        f"{summary_df.iloc[1]['Mean Fire Intensity']:.2f} ± "
        f"{summary_df.iloc[1]['Fire Intensity Std']:.2f}"
    ]
})

publication_table
```

## CELL 15 — Save Publication Table

```python
publication_path = (

    f"{RESULT_DIR}/publication_results_table.csv"
)

publication_table.to_csv(

    publication_path,

    index=False
)

print(f"Saved publication table:\n{publication_path}")
```
