```python
import gymnasium
import numpy as np
import pyrorl

# Set up parameters
num_rows, num_cols = 10, 10
populated_areas = np.array([[1, 2], [4, 8], [6, 4], [8, 7]])
paths = np.array(
    [
        [[1, 0], [1, 1]],
        [[2, 2], [3, 2], [4, 2], [4, 1], [4, 0]],
        [[2, 9], [2, 8], [3, 8]],
        [[5, 8], [6, 8], [6, 9]],
        [[7, 7], [6, 7], [6, 8], [6, 9]],
        [[8, 6], [8, 5], [9, 5]],
        [[8, 5], [9, 5], [7, 5], [7, 4]],
    ],
    dtype=object,
)
paths_to_pops = {
    0: [[1, 2]],
    1: [[1, 2]],
    2: [[4, 8]],
    3: [[4, 8]],
    4: [[8, 7]],
    5: [[8, 7]],
    6: [[6, 4]],
}

# Create environment
kwargs = {
    "num_rows": num_rows,
    "num_cols": num_cols,
    "populated_areas": populated_areas,
    "paths": paths,
    "paths_to_pops": paths_to_pops,
}
env = gymnasium.make("pyrorl/PyroRL-v0", **kwargs)

print("PyroRL environment loaded successfully with all required parameters!")
```

```python
obs = env.reset()

print(type(obs))

try:
    print(obs.shape)
except:
    print("No shape attribute")
```

```python
# Gymnasium reset() returns (observation, info)
obs_array, info = env.reset()

print(f"Observation type: {type(obs_array)}")
print(f"Observation shape: {obs_array.shape}")

# Check compatibility with (7, 32, 32)
if obs_array.shape == (7, 32, 32):
    print("\nPyroRL currently matches the (7, 32, 32) Saudi tensor shape exactly!")
else:
    print(f"\nPyroRL observation shape {obs_array.shape} does NOT naturally match (7, 32, 32).")
    print("To support Saudi tensors, you will likely need to write a custom Gymnasium Observation Wrapper to reshape or pad the state.")
```
