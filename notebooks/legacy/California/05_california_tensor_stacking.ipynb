```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
BASE_32 = "/content/drive/MyDrive/PyroRL_Saudi_Project/datasets/california/grids/32x32"

BASE_64 = "/content/drive/MyDrive/PyroRL_Saudi_Project/datasets/california/grids/64x64"
```

```python
print("32x32 files:")
print(os.listdir(BASE_32))

print("\n64x64 files:")
print(os.listdir(BASE_64))
```

```python
fire_32 = np.load(f"{BASE_32}/fire.npy")

fuel_32 = np.load(f"{BASE_32}/fuel.npy")

wind_x_32 = np.load(f"{BASE_32}/wind_x.npy")

wind_y_32 = np.load(f"{BASE_32}/wind_y.npy")

terrain_32 = np.load(f"{BASE_32}/terrain.npy")

temperature_32 = np.load(f"{BASE_32}/temperature.npy")

humidity_32 = np.load(f"{BASE_32}/humidity.npy")
```

```python
print("fire:", fire_32.shape)

print("fuel:", fuel_32.shape)

print("wind_x:", wind_x_32.shape)

print("wind_y:", wind_y_32.shape)

print("terrain:", terrain_32.shape)

print("temperature:", temperature_32.shape)

print("humidity:", humidity_32.shape)
```

```python
layers_32 = {

    "fire": fire_32,
    "fuel": fuel_32,
    "wind_x": wind_x_32,
    "wind_y": wind_y_32,
    "terrain": terrain_32,
    "temperature": temperature_32,
    "humidity": humidity_32
}

for name, arr in layers_32.items():

    print(
        f"{name}: "
        f"min={arr.min():.4f}, "
        f"max={arr.max():.4f}, "
        f"mean={arr.mean():.4f}"
    )
```

```python
for name, arr in layers_32.items():

    print(
        name,
        np.isnan(arr).sum()
    )
```

```python
state_tensor_32 = np.stack([

    fire_32,        # channel 0
    fuel_32,        # channel 1
    wind_x_32,      # channel 2
    wind_y_32,      # channel 3
    terrain_32,     # channel 4
    temperature_32, # channel 5
    humidity_32     # channel 6
])
```

```python
print(state_tensor_32.shape)
```

```python
state_tensor_32 = state_tensor_32.astype(np.float32)

print(state_tensor_32.dtype)
```

```python
np.save(
    f"{BASE_32}/state_tensor.npy",
    state_tensor_32
)

print("32x32 California tensor saved successfully.")
```

```python
channel_names = [

    "Fire",
    "Fuel",
    "Wind X",
    "Wind Y",
    "Terrain",
    "Temperature",
    "Humidity"
]

fig, axes = plt.subplots(2,4, figsize=(15,8))

axes = axes.flatten()

for i in range(7):

    im = axes[i].imshow(
        state_tensor_32[i]
    )

    axes[i].set_title(channel_names[i])

    plt.colorbar(im, ax=axes[i])

axes[7].axis('off')

plt.tight_layout()

plt.show()
```
